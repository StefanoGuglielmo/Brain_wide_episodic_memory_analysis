```python
from pathlib import Path
import pandas as pd
import spikeinterface.core as si
import sys

REPO_ROOT = Path("..").resolve()
sys.path.append(str(REPO_ROOT))
DATA_DIR = REPO_ROOT / "DATA"
EPHYS_DIR = DATA_DIR / "ephys"

MOUSE_ID = "<MOUSE_ID>"
RECORDING_DIR = EPHYS_DIR / MOUSE_ID

REGION = "<REGION>" #select region to extract
```

```python
# Load the curated spike sorting for the selected mouse and brain region

EXPLORATION_FILE = RECORDING_DIR / "exploration_segments.csv"
LED_FILE = RECORDING_DIR / "LED_info.csv"

SORTING_DIR = RECORDING_DIR / f"{REGION}_curated_sorting/sorting"

sorting_curated = si.load(SORTING_DIR)

print(f"Mouse: {MOUSE_ID}")
print(f"Region: {REGION}")
print(f"Loaded units: {sorting_curated.unit_ids}")
```

```python
# Prepare exploration epochs aligned to the LED synchronization signal

from utils.ephys import prepare_exploration_epochs

exploration_df, epochs_df = prepare_exploration_epochs(
    EXPLORATION_FILE,
    LED_FILE,
)
```

```python
# Extract curated neuronal activity aligned to the start of each exploration epoch

from utils.ephys import extract_exploration_aligned_activity

output_file = (
    RECORDING_DIR
    / f"{MOUSE_ID}_{REGION}_start_interaction.json"
)

extract_exploration_aligned_activity(
    mouse_id=MOUSE_ID,
    region=REGION,
    sorting=sorting_curated,
    epochs_df=epochs_df,
    output_file=output_file,
)
```
